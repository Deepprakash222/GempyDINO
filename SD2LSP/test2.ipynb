```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker

# ── Data ────────────────────────────────────────────────────────────────────
Physical_coordinates_mu = np.array([0.216, 0.25, 0.3, 0.35, 0.384])   # x-axis (swept)

# (100 + 100) samples
# data = {
#     "mu": {
#         "true": None,
#         "det_U": np.array([0.22295986, 0.25398248, 0.30249795, 0.35003419,  0.38243917]),
#         "nuts_U": np.array([0.22304143, 0.25392374, 0.30259679, 0.35015089, 0.38189141]),
#         "det_N": np.array([0.2172819, 0.24999952 , 0.29955761, 0.34928491, 0.38294929]),
#         "nuts_N": np.array([0.21730225, 0.25022241,0.30077007, 0.34946311, 0.38233872]),
#     },
    
# }
# (500 + 500) samples
data = {
    "mu": {
        "true": None,
        "det_U": np.array([0.22295986, 0.25398248, 0.30249795, 0.35003419,  0.38243917]),
        "nuts_U": np.array([0.22331872, 0.25402738, 0.30266583, 0.34988543, 0.38201522]),
        "det_N": np.array([0.2172819, 0.24999952 , 0.29955761, 0.34928491, 0.38294929]),
        "nuts_N": np.array([0.21769874, 0.24971485,0.30045216, 0.34870434, 0.3822478]),
    },
    
}

# ── Training domain ─────────────────────────────────────────────────────────
mu_train_lo = 0.216
mu_train_hi = 0.384



# ── Style ───────────────────────────────────────────────────────────────────
color_nonuniform = "#2a78d6"
color_uniform = "#d95f02"

lw = 2.0
ms = 7

sigma_prior = 0.084


# ── Figure ──────────────────────────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(8, 6))


# ── Ideal μ4 identity line ──────────────────────────────────────────────────
ax.plot(
    Physical_coordinates_mu,
    Physical_coordinates_mu,
    color="black",
    lw=lw,
    ls="-",
    marker="*",
    markersize=ms+4,
    label=r"Ideal $\mu$",
    zorder=5
)


# ── Prior uncertainty band ──────────────────────────────────────────────────
ax.fill_between(
    Physical_coordinates_mu,
    Physical_coordinates_mu - sigma_prior,
    Physical_coordinates_mu + sigma_prior,
    color="black",
    alpha=0.12,
    label=r"Ideal $\pm1\sigma_{\rm prior}$"
)


# ── Uniform NUTS ────────────────────────────────────────────────────────────
ax.plot(
    Physical_coordinates_mu,
    data["mu"]["nuts_U"],
    color=color_uniform,
    lw=lw,
    ls="--",
    marker="o",
    markersize=ms,
    markerfacecolor=color_uniform,
    markeredgecolor="white",
    markeredgewidth=1,
    label="NUTS — Distributed Acquisition"
)


# ── Uniform Deterministic ───────────────────────────────────────────────────
ax.plot(
    Physical_coordinates_mu,
    data["mu"]["det_U"],
    color=color_uniform,
    lw=lw,
    ls="-",
    marker="s",
    markersize=ms,
    markerfacecolor=color_uniform,
    markeredgecolor="white",
    markeredgewidth=1,
    label="GLA — Distributed Acquisition"
)


# ── Non-uniform NUTS ─────────────────────────────────────────────────────────
ax.plot(
    Physical_coordinates_mu,
    data["mu"]["nuts_N"],
    color=color_nonuniform,
    lw=lw,
    ls="--",
    marker="^",
    markersize=ms,
    markerfacecolor=color_nonuniform,
    markeredgecolor="white",
    markeredgewidth=1,
    label="NUTS — Borehole Acquisition"
)


# ── Non-uniform Deterministic ────────────────────────────────────────────────
ax.plot(
    Physical_coordinates_mu,
    data["mu"]["det_N"],
    color=color_nonuniform,
    lw=lw,
    ls="-",
    marker="v",
    markersize=ms,
    markerfacecolor=color_nonuniform,
    markeredgecolor="white",
    markeredgewidth=1,
    label="GLA — Borehole Acquisition"
)


# ── Training range μ4 ───────────────────────────────────────────────────────
ax.axvline(
    mu_train_lo,
    color="gray",
    lw=1.5,
    ls=":",
    label=r"Training range $\mu$"
)

ax.axvline(
    mu_train_hi,
    color="gray",
    lw=1.5,
    ls=":"
)


# ── Formatting ──────────────────────────────────────────────────────────────
ax.set_xlabel(
    r"Physical coordinate of $\mu$ (true)",
    fontsize=12
)

ax.set_ylabel(
    r"Predicted $\mu$ coordinate (posterior mean)",
    fontsize=12
)

ax.set_title(
    r"$\mu$ reconstruction (1% Noise)",
    fontsize=14
)


ax.set_xlim(0.2,0.4)
ax.set_ylim(0.1,0.5)


ax.xaxis.set_major_formatter(
    ticker.FormatStrFormatter("%.2f")
)

ax.yaxis.set_major_formatter(
    ticker.FormatStrFormatter("%.2f")
)


ax.tick_params(labelsize=10)

ax.grid(
    color="gray",
    alpha=0.2,
    linewidth=0.8
)

ax.spines[["top","right"]].set_visible(False)


ax.legend(
    fontsize=9,
    framealpha=0.4,
    loc="best"
)


plt.tight_layout()


plt.savefig(
    "SD2LSP_inversion_1.pdf",
    bbox_inches="tight",
    dpi=150
)

plt.savefig(
    "SD2LSP_inversion_1.png",
    bbox_inches="tight",
    dpi=150
)


plt.show()
```
