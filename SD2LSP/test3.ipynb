```python

import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker

# ── Data ────────────────────────────────────────────────────────────────────
Physical_coordinates_mu = np.array([0.216, 0.25, 0.3, 0.35, 0.384])   # x-axis (swept)


# data = {
#     "mu": {
#         "true": None,
#         "det_U": np.array([0.24601277, 0.2734635, 0.32144089, 0.35742905,  0.38841352]),
#         "nuts_U": np.array([0.24434105, 0.27276516, 0.31384733, 0.35663853, 0.38332978]),
#         "det_N": np.array([0.23031964, 0.26208528, 0.31129242, 0.35617374, 0.38798274]),
#         "nuts_N": np.array([0.22740795, 0.26085037,0.30478779, 0.35378, 0.38286522]),
#     },
    
# }
# (500, 500)
data = {
    "mu": {
        "true": None,
        "det_U": np.array([0.24601277, 0.2734635, 0.32144089, 0.35742905,  0.38841352]),
        "nuts_U": np.array([0.24362672, 0.27180622, 0.31548464, 0.35646543, 0.38374133]),
        "det_N": np.array([0.23031964, 0.26208528, 0.31129242, 0.35617374, 0.38798274]),
        "nuts_N": np.array([0.22996854, 0.26203917,0.30761163, 0.3535718, 0.38179071]),
    },
    
}


# ── Training domain ─────────────────────────────────────────────────────────
mu_train_lo = 0.216
mu_train_hi = 0.384



# ── Style ───────────────────────────────────────────────────────────────────
color_nonuniform = "#2a78d6"
color_uniform = "#d95f02"

lw = 2.0
ms = 7

sigma_prior = 0.084


# ── Figure ──────────────────────────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(8, 6))


# ── Ideal μ4 identity line ──────────────────────────────────────────────────
ax.plot(
    Physical_coordinates_mu,
    Physical_coordinates_mu,
    color="black",
    lw=lw,
    ls="-",
    marker="*",
    markersize=ms+4,
    label=r"Ideal $\mu$",
    zorder=5
)


# ── Prior uncertainty band ──────────────────────────────────────────────────
ax.fill_between(
    Physical_coordinates_mu,
    Physical_coordinates_mu - sigma_prior,
    Physical_coordinates_mu + sigma_prior,
    color="black",
    alpha=0.12,
    label=r"Ideal $\pm1\sigma_{\rm prior}$"
)


# ── Uniform NUTS ────────────────────────────────────────────────────────────
ax.plot(
    Physical_coordinates_mu,
    data["mu"]["nuts_U"],
    color=color_uniform,
    lw=lw,
    ls="--",
    marker="o",
    markersize=ms,
    markerfacecolor=color_uniform,
    markeredgecolor="white",
    markeredgewidth=1,
    label="NUTS — Distributed Acquisition"
)


# ── Uniform Deterministic ───────────────────────────────────────────────────
ax.plot(
    Physical_coordinates_mu,
    data["mu"]["det_U"],
    color=color_uniform,
    lw=lw,
    ls="-",
    marker="s",
    markersize=ms,
    markerfacecolor=color_uniform,
    markeredgecolor="white",
    markeredgewidth=1,
    label="GLA — Distributed Acquisition"
)


# ── Non-uniform NUTS ─────────────────────────────────────────────────────────
ax.plot(
    Physical_coordinates_mu,
    data["mu"]["nuts_N"],
    color=color_nonuniform,
    lw=lw,
    ls="--",
    marker="^",
    markersize=ms,
    markerfacecolor=color_nonuniform,
    markeredgecolor="white",
    markeredgewidth=1,
    label="NUTS — Borehole Acquisition"
)


# ── Non-uniform Deterministic ────────────────────────────────────────────────
ax.plot(
    Physical_coordinates_mu,
    data["mu"]["det_N"],
    color=color_nonuniform,
    lw=lw,
    ls="-",
    marker="v",
    markersize=ms,
    markerfacecolor=color_nonuniform,
    markeredgecolor="white",
    markeredgewidth=1,
    label="GLA — Borehole Acquisition"
)


# ── Training range μ4 ───────────────────────────────────────────────────────
ax.axvline(
    mu_train_lo,
    color="gray",
    lw=1.5,
    ls=":",
    label=r"Training range $\mu$"
)

ax.axvline(
    mu_train_hi,
    color="gray",
    lw=1.5,
    ls=":"
)


# ── Formatting ──────────────────────────────────────────────────────────────
ax.set_xlabel(
    r"Physical coordinate of $\mu$ (true)",
    fontsize=12
)

ax.set_ylabel(
    r"Predicted $\mu$ coordinate (posterior mean)",
    fontsize=12
)

ax.set_title(
    r"$\mu$ reconstruction (5 % Noise)",
    fontsize=14
)


ax.set_xlim(0.2,0.4)
ax.set_ylim(0.1,0.5)


ax.xaxis.set_major_formatter(
    ticker.FormatStrFormatter("%.2f")
)

ax.yaxis.set_major_formatter(
    ticker.FormatStrFormatter("%.2f")
)


ax.tick_params(labelsize=10)

ax.grid(
    color="gray",
    alpha=0.2,
    linewidth=0.8
)

ax.spines[["top","right"]].set_visible(False)


ax.legend(
    fontsize=9,
    framealpha=0.4,
    loc="best"
)


plt.tight_layout()


plt.savefig(
    "SD2LSP_inversion_5.pdf",
    bbox_inches="tight",
    dpi=150
)

plt.savefig(
    "SD2LSP_inversion_5.png",
    bbox_inches="tight",
    dpi=150
)


plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker

# ── Data ────────────────────────────────────────────────────────────────────

mu_train_lo, mu_train_hi = 0.216, 0.384


Relative_function_U = np.array([0.9979, 0.9979, 0.9979, 0.9979, 0.9979])
Relative_function_N = np.array([0.9995, 0.9994, 0.9994, 0.9995, 0.9995])

Relative_Jacobian_U = np.array([0.9065, 0.9855, 0.9759, 0.9867, 0.9855])
Relative_Jacobian_N = np.array([0.9280, 0.9979, 0.9826, 0.9939, 0.9916])
# ── Style ─────────────────────────────────────────────────────────────────────
color_function = "#2ca02c"
color_jacobian = "#d95f02"
color_gradient = "#2a78d6"
lw, ms = 2.0, 8

fig, ax = plt.subplots(figsize=(9, 6.5))

ax.plot(Physical_coordinates_mu, Relative_function_U, color=color_function, lw=lw, ls="--",
        marker="o", markersize=ms, markerfacecolor=color_function, markeredgecolor="white",
        markeredgewidth=1.0, label="Function accuracy — Uniform")
ax.plot(Physical_coordinates_mu, Relative_function_N, color=color_function, lw=lw, ls="-",
        marker="s", markersize=ms, markerfacecolor=color_function, markeredgecolor="white",
        markeredgewidth=1.0, label="Function accuracy — Non-uniform")

ax.plot(Physical_coordinates_mu, Relative_Jacobian_U, color=color_jacobian, lw=lw, ls="--",
        marker="^", markersize=ms, markerfacecolor=color_jacobian, markeredgecolor="white",
        markeredgewidth=1.0, label="Jacobian accuracy — Uniform")
ax.plot(Physical_coordinates_mu, Relative_Jacobian_N, color=color_jacobian, lw=lw, ls="-",
        marker="v", markersize=ms, markerfacecolor=color_jacobian, markeredgecolor="white",
        markeredgewidth=1.0, label="Jacobian accuracy — Non-uniform")


ax.axvline(x=mu_train_lo, color="gray", lw=1.5, ls=":", zorder=1, label="Training range ($\\mu_4$)")
ax.axvline(x=mu_train_hi, color="gray", lw=1.5, ls=":", zorder=1)

ax.set_xlabel(r"Physical coordinate of $\mu_4$ (true)", fontsize=12)
ax.set_ylabel("Relative accuracy", fontsize=12)
ax.set_title("Function, Jacobian, and Gradient-Inversion Accuracy vs. $\\mu_4$", fontsize=13, pad=10)

ax.xaxis.set_major_formatter(ticker.FormatStrFormatter("%.2f"))
ax.yaxis.set_major_formatter(ticker.FormatStrFormatter("%.2f"))
ax.tick_params(labelsize=10)
ax.set_ylim(0.88, 1.005)
ax.grid(color="gray", alpha=0.2, linewidth=0.8)
ax.spines[["top", "right"]].set_visible(False)
ax.legend(fontsize=8.5, framealpha=0.5, loc="lower right", borderpad=0.7)

plt.tight_layout()
plt.savefig("accuracy_metrics_vs_mu4.png", bbox_inches="tight", dpi=150)
plt.savefig("accuracy_metrics_vs_mu4.pdf", bbox_inches="tight", dpi=150)
plt.show()
```
