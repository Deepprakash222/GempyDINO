```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker

# ── Data ────────────────────────────────────────────────────────────────────
Physical_coordinates_mu_4 = np.array([0.15600001, 0.2, 0.3, 0.4, 0.44399999])
mu4_train_lo, mu4_train_hi = 0.15600001, 0.44399999

# Uniform observations
Gradient_accuracy_U_250_250            = np.array([0.8764, 0.8785, 0.8912, 0.9087, 0.9189])
Gradient_accuracy_U_500_J_full         = np.array([0.2677, 0.2327, 0.2354, 0.3146, 0.4491])
Gradient_accuracy_U_500_without_J_full = np.array([0.5815, 0.5665, 0.5403, 0.5584, 0.6198])

# Non-uniform observations
Gradient_accuracy_N_250_250            = np.array([0.8539, 0.8548, 0.8715, 0.8886, 0.8982])
Gradient_accuracy_N_500_J_full         = np.array([0.5034, 0.4802, 0.4753, 0.5187, 0.5679])
Gradient_accuracy_N_500_without_J_full = np.array([0.6921, 0.6936, 0.6910, 0.6748, 0.6693])

# ── Style ─────────────────────────────────────────────────────────────────────
color_U = "#d95f02"   # orange -- uniform
color_N = "#2a78d6"   # blue   -- non-uniform
lw, ms = 2.0, 7

fig, ax = plt.subplots(figsize=(10, 7))

ax.plot(Physical_coordinates_mu_4, Gradient_accuracy_U_250_250, color=color_U, lw=lw, ls="-",
        marker="o", markersize=ms, markerfacecolor=color_U, markeredgecolor="white",
        markeredgewidth=1.0, label=r"Uniform: $\lambda=0\rightarrow1$ (250+250)")
ax.plot(Physical_coordinates_mu_4, Gradient_accuracy_U_500_J_full, color=color_U, lw=lw, ls="--",
        marker="s", markersize=ms, markerfacecolor=color_U, markeredgecolor="white",
        markeredgewidth=1.0, label=r"Uniform: $\lambda=1$ (ep=500)")
ax.plot(Physical_coordinates_mu_4, Gradient_accuracy_U_500_without_J_full, color=color_U, lw=lw, ls=":",
        marker="^", markersize=ms, markerfacecolor=color_U, markeredgecolor="white",
        markeredgewidth=1.0, label=r"Uniform: $\lambda=0$ (ep=500)")

ax.plot(Physical_coordinates_mu_4, Gradient_accuracy_N_250_250, color=color_N, lw=lw, ls="-",
        marker="o", markersize=ms, markerfacecolor=color_N, markeredgecolor="white",
        markeredgewidth=1.0, label=r"Non-uniform: $\lambda=0\rightarrow1$ (250+250)")
ax.plot(Physical_coordinates_mu_4, Gradient_accuracy_N_500_J_full, color=color_N, lw=lw, ls="--",
        marker="s", markersize=ms, markerfacecolor=color_N, markeredgecolor="white",
        markeredgewidth=1.0, label=r"Non-uniform: $\lambda=1$ (ep=500)")
ax.plot(Physical_coordinates_mu_4, Gradient_accuracy_N_500_without_J_full, color=color_N, lw=lw, ls=":",
        marker="^", markersize=ms, markerfacecolor=color_N, markeredgecolor="white",
        markeredgewidth=1.0, label=r"Non-uniform: $\lambda=0$ (ep=500)")

ax.axvline(x=mu4_train_lo, color="gray", lw=1.5, ls=":", zorder=1, label="Training range ($\\mu_4$)")
ax.axvline(x=mu4_train_hi, color="gray", lw=1.5, ls=":", zorder=1)

ax.set_xlabel(r"Physical coordinate of $\mu_4$ (true)", fontsize=12)
ax.set_ylabel("Gradient-inversion accuracy", fontsize=12)
ax.set_title("Gradient-Inversion Accuracy Across Training Strategies", fontsize=13, pad=10)
ax.xaxis.set_major_formatter(ticker.FormatStrFormatter("%.2f"))
ax.yaxis.set_major_formatter(ticker.FormatStrFormatter("%.2f"))
ax.tick_params(labelsize=10)
ax.set_ylim(-0.05, 1.0)
ax.grid(color="gray", alpha=0.2, linewidth=0.8)
ax.spines[["top", "right"]].set_visible(False)
ax.legend(fontsize=9, framealpha=0.5, loc="upper left", borderpad=0.6)

plt.tight_layout()
plt.savefig("gradient_accuracy_training_strategies.png", bbox_inches="tight", dpi=150)
plt.savefig("gradient_accuracy_training_strategies.pdf", bbox_inches="tight", dpi=150)
plt.show()
```
